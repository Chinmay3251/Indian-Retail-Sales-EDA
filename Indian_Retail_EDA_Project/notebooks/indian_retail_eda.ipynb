## Sales breakdowns

```
fig,axs=plt.subplots(1,3,figsize=(18,5))\ndf.groupby('category').net_sales_inr.sum().sort_values().plot.barh(ax=axs[0],title='Category')\ndf.groupby('city').net_sales_inr.sum().sort_values().plot.barh(ax=axs[1],title='City')\ndf.groupby('sales_channel').net_sales_inr.sum().plot.bar(ax=axs[2],title='Channel')\nplt.tight_layout()
```
